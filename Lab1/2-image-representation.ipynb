```python
import cv2 as cv
import numpy as np
```

```python

canvas=np.zeros((300,300,3),dtype="uint8")
green=(0,255,0)
#draw a green line 
cv.line(canvas,(0,0),(300,300),green,5)
cv.imshow("Canvas",canvas)
cv.waitKey(0)
```

```python
red=(0,0,255) #it is in BGR format since we didnt convert yet
#draw a red line 
cv.line(canvas,(0,300),(300,0),red,5)
cv.imshow("Canvas",canvas)
cv.waitKey(0)
```

```python
#draw a rectagle
canvas=np.zeros((300,300,3),dtype="uint8")
cv.rectangle(canvas,(10,30),(100,200),green,5)
cv.imshow("Canvas",canvas)  
cv.waitKey(0)
```

```python
cv.rectangle(canvas,(10,30),(100,200),red,-1)
cv.imshow("Canvas",canvas)  
cv.waitKey(0)
```

```python

canvas=np.zeros((300,300,3),dtype="uint8")
canvas[:] = (255,0,0) #blue background
cv.imshow("Canvas",canvas)  
cv.waitKey(0)
```

```python

canvas=np.zeros((300,300,3),dtype="uint8")
canvas.shape
white=(255,255,255)
```

```python
centerx,centery=canvas.shape[1]//2,canvas.shape[0]//2 #width//2,height//2
print(centerx,centery)
```

```python
for r in range(0,150,25): #(start,stop,step) 150/25=6 circles
    cv.circle(canvas,(centerx,centery),r,white)
cv.imshow("Canvas",canvas)  
cv.waitKey(0)
```

```python
cv.circle(canvas,(centerx,centery),50,(0,255,0),-1)
cv.imshow("Canvas",canvas)  
cv.waitKey(0)
```

```python
for i in range(0,25):
    radius=np.random.randint(5,200)
    color=np.random.randint(0,256,size=(3,)).tolist()   
    pt=np.random.randint(0,300,size=(2,)) #x and y coordinates
    cv.circle(canvas,pt,radius,color,-1)
cv.imshow("Canvas",canvas)  
cv.waitKey(0)
```
