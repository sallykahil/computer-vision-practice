```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
def loadimage():
    blank_image=np.zeros((600,600))
    cv.putText(blank_image,'SALLY',(100,300),cv.FONT_HERSHEY_COMPLEX,4,(255),8,cv.LINE_AA)
    return blank_image
```

```python
def displayimage(image):
    fig=plt.figure(figsize=(8,8))
    ax=fig.add_subplot(111)
    ax.imshow(image,cmap='gray')
    
 
```

```python
img=loadimage()
displayimage(img)
```

### Erosion

```python
kernel=np.ones((5,5),np.uint8) 
erosion=cv.erode(img,kernel,iterations=1)
displayimage(erosion)
```

```python
#increse the iteray=tions to 3
erosion5=cv.erode(img,kernel,iterations=3)
displayimage(erosion5)
```

### Dilation

```python
kernel=np.ones((5,5),np.uint8)
dilation=cv.dilate(img,kernel,iterations=1)
displayimage(dilation)
```

```python

dilation=cv.dilate(img,kernel,iterations=5)
displayimage(dilation)
```

## opening ( erosion+dilation)

```python
img=loadimage()
white_noise=np.random.randint(0,2,size=(600,600))
print(white_noise)
                        
```

```python
white_noise=white_noise*255
displayimage(white_noise)
```

```python
noise_image=white_noise+img
displayimage(noise_image)
```

## opening

```python
opening=cv.morphologyEx(noise_image,cv.MORPH_OPEN,kernel)
displayimage(opening)
```

### closing

```python
img=loadimage()
black_noise=np.random.randint(0,2,size=(600,600))
print(black_noise)
black_noise=black_noise*255
displayimage(black_noise)
```

```python
black_noise_image= black_noise+img
displayimage(black_noise_image)
```

```python
black_noise_image[black_noise_image  == 255] = 0
displayimage(black_noise_image)
```

```python
closing=cv.morphologyEx(black_noise_image,cv.MORPH_CLOSE,kernel)
displayimage(closing)
```

```python
IMG=loadimage()
displayimage(IMG)
```

```python
gradient=cv.morphologyEx(IMG,cv.MORPH_GRADIENT,kernel)
displayimage(gradient)
```
