```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv.imread('../data/image1.png',1)
cv.imshow('Original Image',img)
cv.waitKey(0)
```

```python
color_img=cv.cvtColor(img,cv.COLOR_BGR2GRAY)
cv.imshow('Original Image',color_img)
cv.waitKey(0) 
```

```python
##create prewit FIlter
kernel_x=np.array([[1,0,-1],[1,0,-1],[1,0,-1]])
kernel_y=np.array([[1,1,1],[0,0,0],[-1,-1,-1]])

img_prewit_x=cv.filter2D(color_img,-1,kernel_x)
img_prewit_y=cv.filter2D(color_img,-1,kernel_y)

#hypot bring square root of sum of squares
img_prewit_hypot=np.hypot(img_prewit_x,img_prewit_y)

img_prewit_hypot=img_prewit_hypot[:,:]
img_prewit_hypot=img_prewit_hypot.astype('int')
plt.imshow(img_prewit_hypot,cmap='gray')
```

```python
##create sobel FFilter
kernel_x=np.array([[1,0,-1],[2,0,-2],[1,0,-1]])
kernel_y=np.array([ [1,2,1],[0,0,0],[-1,-2,-1]])    

img_sobel_x=cv.filter2D(color_img,-1,kernel_x)
img_sobel_y=cv.filter2D(color_img,-1,kernel_y)

#hypot bring square root of sum of squares
img_sobel_hypot=np.hypot(img_sobel_x,img_sobel_y)   
img_sobel_hypot=img_sobel_hypot[:,:]
img_sobel_hypot=img_sobel_hypot.astype('int')
plt.imshow(img_sobel_hypot,cmap='gray')
```

```python
#sobel show edges better than prewit
```

```python
#canny edge detection
image =img
(h,w)= image.shape[:2]
gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
blurred = cv.GaussianBlur(gray, (5, 5), 0)

edged = cv.Canny(blurred, 80, 150)
fig, ax = plt.subplots(1, 2, figsize=(18, 18))
ax[0].imshow(gray, cmap='gray')
ax[0].set_title('Gray Image')
ax[1].imshow(edged, cmap='gray')
ax[1].set_title('Canny Edges')
ax[0].axis('off')
ax[1].axis('off')
```
