```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt
```

```python
image=cv.imread('../data/flat_chessboard.png')
plt.imshow(image)
```

## calculate the grid


```python
found ,corners=cv.findChessboardCorners(image,(7,7),None)
if found:
    flat_chess_copy = cv.drawChessboardCorners(image,(7,7),corners,found) 
print(f"Chessboard found: {flat_chess_copy}")
plt.imshow(flat_chess_copy)
```
