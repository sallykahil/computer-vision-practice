```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt
```

```python
image=cv.imread('../data/sammy_face.jpg')
plt.imshow(image) 
```

```python
edges=cv.Canny(image=image,threshold1=127,threshold2=127)
plt.imshow(edges)
```

```python
#first we need to convert to grey 
gray=cv.cvtColor(image,cv.COLOR_BGR2GRAY)
plt.imshow(gray,cmap='gray')
```

## Choosing Threshold

```python
med_value=np.median(gray)
lower=int(max(0,0.7*med_value)) #first 70% of median
upper=int(min(255,1.3*med_value)) #last 30 above median
```

```python
edges =cv.Canny(image=gray,threshold1=lower,threshold2=upper)
plt.imshow(edges,cmap='gray')
```

```python
#we need to blurr to remove noise
gray_real_chess=cv.blur(src=gray,ksize=(5,5))
edges=cv.Canny(image=gray_real_chess,threshold1=lower,threshold2=upper)
plt.imshow(edges,cmap='gray')
```
