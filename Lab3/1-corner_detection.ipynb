```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
flat_chess=cv.imread('../data/flat_chessboard.png')
flat_chess=cv.cvtColor(flat_chess,cv.COLOR_BGR2RGB)
plt.imshow(flat_chess)
```

```python
gray_flat_chess=cv.cvtColor(flat_chess,cv.COLOR_RGB2GRAY)
plt.imshow(gray_flat_chess,cmap='gray')
```

```python
real_chess=cv.imread('../data/real_chessboard.jpg')
real_chess=cv.cvtColor(real_chess,cv.COLOR_BGR2RGB)
plt.imshow(real_chess)
```

```python
gray_real_chess=cv.cvtColor(real_chess,cv.COLOR_RGB2GRAY)
plt.imshow(gray_real_chess,cmap='gray')
```

### HAARIS corner detection

```python
grey=np.float32(gray_flat_chess)

dst=cv.cornerHarris(src=grey,blockSize=2,ksize=3,k=0.04)
#dilate to mark the corners
dst=cv.dilate(dst,None)
#threshold to get the corners
flat_chess[dst>0.01*dst.max()]=[255,0,0]
plt.imshow(flat_chess)
```

```python
gray_real_chess=np.float32(gray_real_chess)

dst=cv.cornerHarris(src=gray_real_chess,blockSize=2,ksize=3,k=0.04)
#dilate to mark the corners
dst=cv.dilate(dst,None) 
real_chess[dst>0.01*dst.max()]=[255,0,0]
#threshold to get the corners 
plt.imshow(real_chess)
```
